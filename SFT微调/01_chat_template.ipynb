```python
from transformers import AutoTokenizer, AutoModelForCausalLM
Mytokenizer = AutoTokenizer.from_pretrained("./model/Qwen3-0.6B-Base")
```

```python
message = [
    {"role":"user","content":"请用中文介绍一下你自己。"},
    {"role":"assistant","content":"我是一个由人工智能驱动的语言模型，能够理解和生成自然语言。我可以回答问题、提供信息、进行对话，并帮助用户完成各种任务。我的目标是为用户提供有用的帮助和支持。"}
]
print(message)
```

```python
new_message = Mytokenizer.apply_chat_template(message,tokenize = False)
print(new_message)
```

```python
message2 = [
    {"role":"system","content":"你是一个有帮助的助手。"},
    {"role":"user","content":"请用中文介绍一下你自己。"},
]
message_generated = Mytokenizer.apply_chat_template(message2,tokenize = False,add_generation_prompt=True)
print(message_generated)
```

```python
message3 = [
    {"role":"system","content":"你是一个有帮助的助手。"},
    {"role":"user","content":"请用中文介绍一下你自己。"},
]
message_generated_No_Think = Mytokenizer.apply_chat_template(message3,tokenize = False,add_generation_prompt=True,enable_thinking=False)
print(message_generated_No_Think)
```
